# 1. 데이터 준비
README 설치 절차를 먼저 실행하고 LLM BERTopic 커널을 선택하세요. 설정 파일의 분할과 전처리를 사용합니다. 최초 실행은 데이터를 다운로드합니다.

```python
from pathlib import Path
from llm_bertopic.config import load_config, effective_config

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").is_file())
print("Repository:", ROOT)
```

```python
# minimal 실험은 CONFIG를 해당 configs/minimal_*.yaml로 변경하세요.
# 전처리: full / minimal_keep_stopwords / minimal_remove_stopwords
CONFIG = ROOT / "configs/research.yaml"
DATASETS = ["newsgroup20", "bbc", "imdb"]
# 빠른 확인: --set에 해당하는 override를 추가할 수 있습니다.
cfg = load_config(CONFIG, root=ROOT)
cfg["experiment"]["datasets"] = DATASETS
```

```python
from llm_bertopic.data import prepare_dataset

corpora = {}
for dataset in DATASETS:
    corpus = prepare_dataset(cfg, dataset)
    corpora[dataset] = corpus
    print(dataset, len(corpus["texts"]), "documents; dropped", corpus["metadata"]["dropped_count"])
    print("Corpus hash:", corpus["metadata"]["corpus_sha256"])
```
